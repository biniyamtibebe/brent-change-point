```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pymc as pm
import arviz as az
from statsmodels.tsa.stattools import adfuller

# Load and prepare
df = pd.read_csv(r'c:\Users\hp\Pictures\brent-change-point\brent-change-point\data\raw\BrentOilPrices.csv')
df['Date'] = pd.to_datetime(df['Date'], format='%d-%b-%y')
df = df.sort_values('Date').set_index('Date')
df = df.loc['2010-01-01':].copy()   # Focus on relevant period (~4600 obs)

print("Data shape:", df.shape)
print("ADF on Price (non-stationary expected):", adfuller(df['Price'])[1])
print("ADF on Log Returns:", adfuller(np.log(df['Price']).diff().dropna())[1])

# EDA plots (save as images for report)
df['Price'].plot(figsize=(14,7), title='Brent Oil Price (2010-2022)')
plt.axvline(pd.to_datetime('2014-11-27'), color='red', linestyle='--', alpha=0.7)
plt.savefig('reports/price_series.png')

df['LogReturn'] = np.log(df['Price']).diff()
df['LogReturn'].plot(title='Log Returns + Volatility Clustering')
```
