```python
import requests
import json

# Test 1: Get price data
response = requests.get("http://127.0.0.1:5000/api/prices", 
                       params={"start_date": "2020-01-01", "end_date": "2020-06-01"})
print(f"Status: {response.status_code}")
print(f"Data points: {response.json()['count']}")
print(f"First price: {response.json()['data'][0]}")

# Test 2: Get events
response = requests.get("http://127.0.0.1:5000/api/events")
events = response.json()['events']
print(f"\nNumber of events: {len(events)}")
for event in events[:3]:  # Show first 3
    print(f"- {event['date']}: {event['description'][:50]}...")
```
